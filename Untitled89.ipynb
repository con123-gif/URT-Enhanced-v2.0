```python
import numpy as np
import math
DELTA_STAR = 0.1475
```

```python
def adaptive_urt(x):
    x = (x - np.mean(x)) / (np.std(x) + 1e-10)
    acf = np.correlate(x, x, 'full')[-len(x):]
    acf /= acf[0]
    d = np.where(acf < np.exp(-1))[0]
    tau_D = d[0] if len(d)>0 else len(x)//10
    D = 1 + 2/(1+np.exp(-tau_D/10))
    D = np.clip(D,1,5)
    k = min(20, len(x)//50)
    v = [np.var(x[i::k]) for i in range(k)]
    tau = 2 + 0.5*np.mean(v)/(np.std(v)+1e-10) if len(v)>1 else 2.0
    tau = np.clip(tau,1.5,3.5)
    delta = (D-1)*(tau-2)
    delta = np.clip(delta,0.01,1.0)
    for n in range(30):
        kappa = delta**2/(1+delta**2)
        alpha = 0.5 * np.exp(-n/8)
        delta -= alpha * (delta - DELTA_STAR) * (1 + kappa)
        delta = np.clip(delta,0.001,0.5)
    return delta
```

```python
def logistic(r, x): return r*x*(1-x)
x = [0.5]
for _ in range(5000): x.append(logistic(3.9, x[-1]))
print("Natural chaos delta =", round(adaptive_urt(np.array(x)),6))
r = 3.9
for i in range(5000):
    x.append(logistic(r, x[-1]))
    if i>200 and i%100==0:
        recent = np.array(x[-200:])
        r -= 0.01 * (adaptive_urt(recent) - DELTA_STAR)
        r = np.clip(r,3.5,4.0)
print("Controlled delta =", round(adaptive_urt(np.array(x)),6))
```
