```python
from google.colab import drive
import os
import shutil
from hashlib import md5

# ====================== MOUNT GOOGLE DRIVE ======================
print("Mounting Google Drive...")
drive.mount('/content/drive', force_remount=False)
print("Drive mounted successfully!\n")

# ====================== PATHS ======================
drive_colab_path = '/content/drive/MyDrive/Colab Notebooks'   # Source (your Colab files in Drive)
local_colab_path = '/content/Colab_Notebooks'                 # Destination (local session)

os.makedirs(local_colab_path, exist_ok=True)

# ====================== HELPER: Calculate file hash ======================
def get_file_hash(filepath):
    hash_md5 = md5()
    with open(filepath, "rb") as f:
        for chunk in iter(lambda: f.read(4096), b""):
            hash_md5.update(chunk)
    return hash_md5.hexdigest()

# ====================== MAIN FUNCTION: Copy without duplicates ======================
def sync_colab_files(src, dst):
    copied = 0
    skipped = 0
    conflicts = 0

    print(f"Scanning: {src}\n")

    for root, dirs, files in os.walk(src):
        rel_path = os.path.relpath(root, src)
        dst_root = os.path.join(dst, rel_path) if rel_path != '.' else dst
        os.makedirs(dst_root, exist_ok=True)

        for file in files:
            if file.endswith('.ipynb'):
                src_file = os.path.join(root, file)
                dst_file = os.path.join(dst_root, file)

                if os.path.exists(dst_file):
                    if get_file_hash(src_file) == get_file_hash(dst_file):
                        skipped += 1
                        continue  # Exact duplicate → skip
                    else:
                        print(f"⚠️  Conflict (different content): {file} — skipping")
                        conflicts += 1
                        continue

                shutil.copy2(src_file, dst_file)
                copied += 1
                print(f"✅ Copied: {file}")

    print(f"\n{'='*50}")
    print(f"✅ Done! Copied: {copied} files")
    print(f"⏭️  Skipped (exact duplicates): {skipped}")
    print(f"⚠️  Skipped (name conflict, different content): {conflicts}")
    print(f"📁 All your Colab files are now in: {dst}")
    print(f"{'='*50}")

# ====================== RUN IT ======================
sync_colab_files(drive_colab_path, local_colab_path)
```

```python
import os
import nbformat
from nbformat import v4 as nbf

# ====================== SETTINGS ======================
source_folder = '/content/Colab_Notebooks'
output_file = '/content/All_My_Colab_Notebooks_Combined.ipynb'

# ====================== CREATE NEW NOTEBOOK ======================
combined = nbf.new_notebook()
combined.cells = []

# Add a nice header
header = nbf.new_markdown_cell(
    "# 📚 ALL MY COLAB NOTEBOOKS COMBINED\n\n"
    f"**Total notebooks merged:** 165\n\n"
    "Each section below comes from one original notebook."
)
combined.cells.append(header)

# ====================== MERGE ALL NOTEBOOKS ======================
print("Merging all notebooks...\n")

count = 0
for filename in sorted(os.listdir(source_folder)):
    if filename.endswith('.ipynb'):
        filepath = os.path.join(source_folder, filename)
        try:
            with open(filepath, 'r', encoding='utf-8') as f:
                nb = nbformat.read(f, as_version=4)

            # Add separator
            separator = nbf.new_markdown_cell(f"## 📄 From: {filename}")
            combined.cells.append(separator)

            # Add all cells from this notebook
            combined.cells.extend(nb.cells)

            count += 1
            print(f"✅ Merged: {filename}")

        except Exception as e:
            print(f"⚠️  Skipped {filename}: {e}")

# ====================== SAVE ======================
with open(output_file, 'w', encoding='utf-8') as f:
    nbformat.write(combined, f)

print(f"\n{'='*60}")
print(f"🎉 DONE! All notebooks merged into ONE file:")
print(f"📁 {output_file}")
print(f"📊 Total notebooks merged: {count}")
print(f"{'='*60}")
```

```python
import os
import json

source_folder = '/content/Colab_Notebooks'
output_file = '/content/All_My_Code_Combined.py'

print("Creating one single .py file from all notebooks...\n")

all_code = []

# Add a header
all_code.append("#" * 80)
all_code.append("# ALL MY COLAB NOTEBOOKS — COMBINED INTO ONE FILE")
all_code.append("# Total notebooks: 165")
all_code.append("#" * 80)
all_code.append("")

for filename in sorted(os.listdir(source_folder)):
    if filename.endswith('.ipynb'):
        filepath = os.path.join(source_folder, filename)
        try:
            with open(filepath, 'r', encoding='utf-8') as f:
                nb = json.load(f)

            # Add separator for this notebook
            all_code.append("#" * 80)
            all_code.append(f"# 📄 FROM: {filename}")
            all_code.append("#" * 80)
            all_code.append("")

            # Extract only code cells
            for cell in nb.get('cells', []):
                if cell.get('cell_type') == 'code':
                    source = cell.get('source', [])
                    if isinstance(source, list):
                        code_text = ''.join(source)
                    else:
                        code_text = source

                    if code_text.strip():  # skip empty cells
                        all_code.append(code_text)
                        all_code.append("")  # blank line between cells

            all_code.append("")  # extra space after each notebook

        except Exception as e:
            print(f"⚠️  Skipped {filename}: {e}")

# Write everything to one file
with open(output_file, 'w', encoding='utf-8') as f:
    f.write('\n'.join(all_code))

print(f"\n{'='*70}")
print(f"✅ DONE! Everything is now in ONE file:")
print(f"📁 {output_file}")
print(f"{'='*70}")
print("You can download it with:")
print("from google.colab import files")
print("files.download('/content/All_My_Code_Combined.py')")
```
