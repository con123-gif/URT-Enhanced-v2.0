```python
from google.colab import drive, files
drive.mount('/content/drive')

import os

SOURCE = '/content/drive/MyDrive'   # narrow this to your research folder
EXTS   = {'.ipynb', '.py', '.csv', '.md', '.json', '.txt'}  # None = everything
SKIP   = {'.git', '__pycache__', '.ipynb_checkpoints'}

paths = []
for root, dirs, fs in os.walk(SOURCE):
    dirs[:] = [d for d in dirs if d not in SKIP]
    for f in fs:
        if EXTS is None or os.path.splitext(f)[1].lower() in EXTS:
            paths.append(os.path.join(root, f))

print(f'{len(paths)} files queued')

for i, p in enumerate(paths, 1):
    print(f'{i}/{len(paths)}  {os.path.basename(p)}')
    files.download(p)
```
