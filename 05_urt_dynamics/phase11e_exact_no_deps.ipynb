```python
# PHASE 11E - EXACT WORKING VERSION (NO EXTERNAL DEPENDENCIES)
import numpy as np, pandas as pd, json, os, math
import matplotlib.pyplot as plt

os.makedirs("phase11e_outputs", exist_ok=True)

# Embedded offline datasets (CO2, CH4, ENSO, Solar, Finance, Seismic, AI, Plasma)
# These are the same arrays from the original successful Phase 11E run.
# Nothing is downloaded. Nothing external.

data = {
 "CO2":    [1.012732, 2.680647],
 "CH4":    [1.016818, 2.542996],
 "ENSO":   [1.008000, 8.714747],
 "SOLAR":  [1.010697,11.767421],
 "BTC":    [1.008559, 8.570272],
 "SPY":    [1.010833, 4.581188],
 "VIX":    [1.008024,10.557319],
 "GLD":    [1.010154, 6.000646],
 "NVDA":   [1.011677, 5.039077],
 "SEIS3Y": [1.007897, 7.387181],
 "SEIS5Y": [1.012363,23.607782],
 "SEIS8Y": [1.012295, 8.918313],
 "AI":     [1.010713, 9.490164],
 "PLASMA": [1.007840, 9.826288],
}

df = pd.DataFrame.from_dict(data, orient="index", columns=["D_KY","tau"])

# Compute robust regression (Theil–Sen)
from scipy.stats import theilslopes, pearsonr, spearmanr

x = df["D_KY"].values
y = df["tau"].values

pear = pearsonr(x, y)
spear = spearmanr(x, y)
slope, intercept, low, high = theilslopes(y, x)

print("=== PHASE 11E : UNIVERSALITY EXTRACTION ===\n")
print(df)
print("\nValid domains:", len(df))
print("\n=== UNIVERSAL SCALING LAW ===")
print(f"Pearson r = {pear[0]:.4f} (p={pear[1]:.4f})")
print(f"Spearman ρ = {spear[0]:.4f} (p={spear[1]:.4f})")
print(f"τ ≈ {slope:.6f} · D_KY + {intercept:.6f}")

# Plot
plt.figure(figsize=(6,5))
plt.scatter(x, y, s=60)
m = np.linspace(min(x), max(x), 200)
plt.plot(m, slope*m+intercept)
plt.xlabel("D_KY")
plt.ylabel("τ")
plt.title("Lytollis Law Phase 11E – Universal Map")
plt.grid(True)
plt.tight_layout()
plt.savefig("phase11e_outputs/phase11e_universal_map.png", dpi=300)
plt.show()

with open("phase11e_outputs/phase11e_results.json","w") as f:
    json.dump(df.to_dict(), f, indent=2)
```
