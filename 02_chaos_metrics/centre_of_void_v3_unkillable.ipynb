```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Define Constants and Range ---
δ_center = 0.1475
band = 0.015
δ_low = δ_center - band
δ_high = δ_center + band

# Observed CODATA/Experimental Values
obs_alpha = 137.035999206
obs_mass = 1836.15267344
obs_weak = 0.23122 # Using the commonly cited value

# Define a range for δ that covers the band and a little more
delta = np.linspace(δ_center - 0.05, δ_center + 0.05, 500)
# Ensure delta does not hit zero for the inverse formulas
delta = delta[delta > 0.001]

# --- 2. Calculate Predictions Across the Range ---
# 1/α Prediction: p_alpha = 1 + 1/δ
p_alpha = 1 + 1 / delta

# Proton/Electron Mass Prediction: p_mass = 12 * π² / δ
p_mass = 12 * np.pi**2 / delta

# Weak Angle Prediction: p_weak = δ / (π - 2)
p_weak = delta / (np.pi - 2)

# --- 3. Plotting ---
fig, axes = plt.subplots(3, 1, figsize=(10, 15))
plt.suptitle(r'Sensitivity of Predicted Constants to $\delta$', fontsize=16)

# --- Plot 1: Inverse Fine-Structure Constant (1/α) ---
ax = axes[0]
ax.plot(delta, p_alpha, label=r'$1/\alpha = 1 + 1/\delta$', color='blue')
ax.axhline(obs_alpha, color='red', linestyle='--', label='Observed $1/\alpha$')
ax.axvspan(δ_low, δ_high, color='green', alpha=0.1, label=r'Tuning Band $\delta \pm 0.015$')
ax.axvline(δ_center, color='black', linestyle=':', label=r'Centre $\delta_{\star}$')
ax.set_ylabel(r'Predicted $1/\alpha$ Value', fontsize=12)
ax.set_title(r'$1/\alpha$ Prediction vs. $\delta$ (Highly Sensitive)', fontsize=14)
ax.legend(loc='upper right')
ax.grid(True, linestyle='dotted', alpha=0.6)

# --- Plot 2: Proton/Electron Mass Ratio (mp/me) ---
ax = axes[1]
ax.plot(delta, p_mass, label=r'$m_p/m_e = 12\pi^2/\delta$', color='purple')
ax.axhline(obs_mass, color='red', linestyle='--', label='Observed $m_p/m_e$')
ax.axvspan(δ_low, δ_high, color='green', alpha=0.1)
ax.axvline(δ_center, color='black', linestyle=':')
ax.set_ylabel(r'Predicted $m_p/m_e$ Value', fontsize=12)
ax.set_title(r'Proton/Electron Mass Prediction vs. $\delta$ (Highly Sensitive)', fontsize=14)
ax.legend(loc='upper right')
ax.grid(True, linestyle='dotted', alpha=0.6)

# --- Plot 3: Weak Mixing Angle (sin²θ_W) ---
ax = axes[2]
ax.plot(delta, p_weak, label=r'$\sin^2\theta_W = \delta/(\pi-2)$', color='orange')
ax.axhline(obs_weak, color='red', linestyle='--', label='Observed $\sin^2\theta_W$')
ax.axvspan(δ_low, δ_high, color='green', alpha=0.1)
ax.axvline(δ_center, color='black', linestyle=':')
ax.set_xlabel(r'Constant $\delta$ (Centre = 0.1475)', fontsize=12)
ax.set_ylabel(r'Predicted $\sin^2\theta_W$ Value', fontsize=12)
ax.set_title(r'$\sin^2\theta_W$ Prediction vs. $\delta$ (Linearly Sensitive)', fontsize=14)
ax.legend(loc='upper left')
ax.grid(True, linestyle='dotted', alpha=0.6)

plt.tight_layout(rect=[0, 0.03, 1, 0.98])
plt.show()
```
